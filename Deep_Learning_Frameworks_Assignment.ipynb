Ques - 1- What is TensorFlow 2.0, and how is it different from TensorFlow 1.x?

--> TensorFlow 2.0 is a major upgrade from TensorFlow 1.x, bringing significant changes to the API, eager execution, and overall development workflow. Here are some key differences:

*   **Eager Execution by Default:** In TensorFlow 1.x, you had to explicitly enable eager execution. TensorFlow 2.0 makes eager execution the default, allowing for more intuitive and interactive development similar to Python. This means operations are executed immediately rather than building a static graph.

*   **Simplified API:** TensorFlow 2.0 streamlines the API, removing redundant functions and making it more consistent. Keras is the recommended high-level API for building models, and it's deeply integrated into TensorFlow 2.0.

*   **No More Sessions:** TensorFlow 1.x relied heavily on sessions to run graphs. TensorFlow 2.0 eliminates the need for sessions in eager execution, simplifying the code and making it easier to debug.

*   **`@tf.function` for Performance:** While eager execution is great for development, graphs are still crucial for performance and deployment. TensorFlow 2.0 introduces the `@tf.function` decorator, which allows you to convert Python functions into TensorFlow graphs for improved performance.

*   **Variables and Resource Management:** Variable management is simplified in TensorFlow 2.0. Variables are tracked automatically, and there's no need for explicit initialization or fetching within a session.

*   **SavedModel:** TensorFlow 2.0 uses the SavedModel format as the standard for saving and loading models, making it easier to deploy models in various environments.

- **Deployment:** TensorFlow has a more mature ecosystem for deployment to various platforms (mobile, web, etc.) through TensorFlow Lite and TensorFlow.js. PyTorch also has deployment options, but TensorFlow's ecosystem is generally considered more extensive.
- **API:** While both have high-level APIs (Keras for TensorFlow, and torch.nn for PyTorch), their underlying structures and lower-level APIs differ.

Ques -16. What are some common activation functions used in neural networks?

-->
Activation functions are crucial components of neural networks as they introduce non-linearity, allowing the network to learn complex patterns. Here are some common activation functions:

*   **ReLU (Rectified Linear Unit)**: This is one of the most popular activation functions. It outputs the input directly if it's positive, otherwise, it outputs zero. It's computationally efficient and helps to mitigate the vanishing gradient problem.
    *   Formula: `f(x) = max(0, x)`

*   **Sigmoid**: This function squashes the input to a range between 0 and 1. It was historically used in the output layer for binary classification problems but is less common in hidden layers due to the vanishing gradient problem.
    *   Formula: `f(x) = 1 / (1 + e^(-x))`

*   **Tanh (Hyperbolic Tangent)**: Similar to the sigmoid function, but it squashes the input to a range between -1 and 1. It is also less common in hidden layers of deep networks due to vanishing gradients.
    *   Formula: `f(x) = (e^x - e^(-x)) / (e^x + e^(-x))`

*   **Leaky ReLU**: A variation of ReLU that allows a small, non-zero gradient when the input is negative. This can help to address the "dying ReLU" problem where neurons become inactive.
    *   Formula: `f(x) = max(0.01x, x)` (the slope for negative values can vary)

*   **Softmax**: This function is typically used in the output layer of multi-class classification problems. It converts a vector of numbers into a probability distribution, where the sum of the outputs is 1.
    *   Formula: `Softmax(xi) = e^xi / sum(e^j for all j)`

*   **Swish**: A relatively newer activation function that is a smooth, non-monotonic function. It has been shown to perform well in some deep learning models.
    *   Formula: `f(x) = x * sigmoid(x)`


Ques -17. What is the difference between torch.nn.Module and torch.nn.Sequential in PyTorch?

-->
`torch.nn.Module` and `torch.nn.Sequential` are both used to build neural networks in PyTorch, but they serve different purposes:

*   **`torch.nn.Module`**: This is the base class for all neural network modules in PyTorch. When you create a custom neural network, you will typically define a class that inherits from `torch.nn.Module`. This class allows you to define the layers of your network in the `__init__` method and specify the forward pass (how data flows through the network) in the `forward` method. `torch.nn.Module` provides essential functionalities like tracking parameters, moving the module to different devices (CPU/GPU), and managing submodules.

*   **`torch.nn.Sequential`**: This is a container module that allows you to stack layers in a sequential manner. It's a convenient way to build simple feedforward networks where the output of one layer is the input to the next. You can pass a list of modules (layers) to `torch.nn.Sequential`, and it will automatically connect them in the order they are provided. This simplifies the `forward` method, as you don't need to explicitly define the data flow between layers.

In summary:

*   Use `torch.nn.Module` when you need more flexibility in defining your network architecture, such as networks with multiple inputs/outputs, skip connections, or more complex data flow.
*   Use `torch.nn.Sequential` for building simple, feedforward networks where layers are applied in a linear sequence.

```python
# Ques -8. How do you implement a custom loss function in PyTorch?

import torch as tr
import torch.nn as nn
import torch.nn.functional as F

# Option 1: Create a class inheriting from nn.Module
class CustomMSELoss(nn.Module):
    def __init__(self):
        super(CustomMSELoss, self).__init__()

    def forward(self, input, target):
        # Calculate Mean Squared Error
        loss = tr.mean((input - target)**2)
        return loss

# Option 2: Create a simple function
def custom_mae_loss(input, target):
    # Calculate Mean Absolute Error
    loss = tr.mean(tr.abs(input - target))
    return loss

# Example usage:
# Assuming 'output' is the prediction from your model and 'target' is the ground truth
output = tr.randn(10, 1)  # Example output tensor
target = tr.randn(10, 1)  # Example target tensor

# Using the class-based custom loss
custom_mse = CustomMSELoss()
loss_mse = custom_mse(output, target)
print(f"Custom MSE Loss (class): {loss_mse.item()}")

# Using the function-based custom loss
loss_mae = custom_mae_loss(output, target)
print(f"Custom MAE Loss (function): {loss_mae.item()}")
```
